## Key Concepts — IOC-2 Reference

### IOC-2.C — How Attackers Get In

| Attack | Type | How it works |
|---|---|---|
| Phishing | Credential theft | Fake emails/sites clone trusted brands to steal login credentials |
| Keylogging | Surveillance | Malware records every keystroke and sends data to attacker |
| Rogue access point | Network intercept | Fake Wi-Fi hotspot intercepts all unencrypted traffic |
| Malicious links | Drive-by install | Clicking installs malware or redirects to phishing pages |
| Malware & viruses | System compromise | Damages systems; viruses self-replicate by attaching to programs |

### IOC-2.B — Defense in Depth

1. **Strong passwords** — passphrases beat short complex passwords
2. **MFA** — something you know + something you have + something you are
3. **Encryption** — symmetric (1 key) vs public key (2 keys)
4. **Software updates** — patch known vulnerabilities before attackers exploit them
5. **Antivirus & permissions** — scan for malware; review app permissions regularly

### IOC-2.B.5 — Encryption Types

| Type | Keys | How it works |
|---|---|---|
| Symmetric | 1 shared key | Same key encrypts and decrypts — fast but requires secure key exchange |
| Public key | 2 keys (public + private) | Public key encrypts; only the private key decrypts |
| Certificate Authority (CA) | — | Issues digital certificates that verify public key ownership (enables HTTPS) |
